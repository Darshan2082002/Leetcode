```python
def two_pointer(arr,target):
    n=len(arr)
    left=0
    right=n-1
    while left<right:
        if sum(arr[left]+arr[right])==target:
            return True 
        elif sum(arr[left]+arr[right])<target:
            left+=1
        else:
            right-=1
    return False 


    
if __name__=="__main__":
    arr=[1,2,3,4,6]
    target=7
    print(two_pointer(arr,target))
```
