```python
class Solution:
    def removeElement(self, nums, val) -> int:
        n=len(nums)
        left=0
        right=n-1
       
        while left<=right:
            if nums[left]==val:
                nums[left]=nums[right]
                right-=1
            else:
                left+=1
        return left
```

```python
solution = Solution()
solution.removeElement([3,2,2,3], 3)
```
