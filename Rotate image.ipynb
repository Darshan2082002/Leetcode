```python
class solution:
    def rotate (self,matrix:list[list[int]])->None:
        n=len(matrix)
        for i in range(n):
            for  j in range(i,n):
                matrix[i][j],matrix[j][i]=matrix[j][i],matrix[i][j]
        for row in matrix:
            row.reverse()
        return matrix
if __name__=="__main__":
    matrix=[[1,2,3],[4,5,6],[7,8,9]]
    sol=solution()
    print(sol.rotate(matrix))
    
```
