```python
class Solution:
    def removeElement(self, nums, val) -> int:
        n=len(nums)
        left=0
        right=n-1
       
        while left<=right:
            if nums[left]==val:
                nums[left]=nums[right]
                right-=1
            else:
                left+=1
        return left
```

```python
solution = Solution()
solution.removeElement([3,2,2,3], 3)
```

```python
#Backtracking implementation on subset sum problem 
def subset_sum(arr):
    result=[]
    def backtrack_helper(index,arr,result,temp=[]):
        if index==len(arr):
            result.append(temp[:])
            return 
        for i in range(index,len(arr)):
            temp.append(arr[i])
            backtrack_helper(index+1,arr,result,temp)
            temp.pop()
            backtrack_helper(index+1,arr,result)
            arr[index],arr[i]=arr[i],arr[index]
    backtrack_helper(0,arr,result)
    return result[::-1]


if __name__=="__main__":
    arr=[1,2,3]
    print(subset_sum(arr))
```
